```python
# Task 1: String Concatenation
name=input("enter your name:")
string1="hello"+ name
print("string1")

string3=("welcome to python programming")
full_string=string1+string3
print(full_string)

# Task 2: String Slicing and Indexing
print("\na first charactor:",full_string[0])
print("b. last charactor:",full_string[-1])
print("c.first 5 charactor:",full_string[:5])
print("d last 11 charactor:",full_string[-11:])
print("e reversed string:",full_string[-1:])
# Use slicing and print the word “Python” from the existing string.

start_index = full_string.find("Python")
print("f. Extracted word:", full_string[start_index:start_index + 6])
```

```python
# String Methods

strM="PYTHON BEGINNER TUTTORIAL"

# a. Uppercase
print("a. uppercase:",strM.upper())

# b. Lowercase
print("b. lowercase:",strM.lower())

# c. Capitalize
print("c.  capitalize:",strM.capitalize())

# d. Count 't'
print("d count of t:",strM.lower().count('t'))

# e. Replace "Python" with "Data Analytics"
strM_replaced= strM.replace("python","data analysis")
print("e. replace string:",strM_replaced)




```

```python
# Python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

# a. Concatenate

t_combine=tuple1+ tuple2
print("a compined tuple:",t_combine)

# b. Repeat 3 times
print("b repeated 3 time:",t_combine *3)

# c. 3rd element (index 2)
print("c. 3rd element:", t_combine[2])

# d. First 3 elements
print("d. First 3 elements:", t_combine[:3])

# e. Last 3 elements
print("e. Last 3 elements:", t_combine[-3:])
```
